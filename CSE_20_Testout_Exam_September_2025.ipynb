## Question 1
Write a function `split_list()` that, given a list `l`, splits `l` into two smaller lists: the list consisting of the even-position elements of `l`, and the list consisting of the odd-position elements of `l`. For instance,

```
split_list(['a', 'b', 'c', 'd', 'e'])
```

should return two lists:

```
['a', 'c', 'e'], ['b', 'd']
```

````python
def split_list(l):
    """Splits the list l into the two list of elements appearing
    in the even positions of l, and in the odd positions of l."""
    ### YOUR SOLUTION HERE
    l_odd = []
    l_even = []

    for i in range(len(l)):
      if(i % 2 == 0):
        l_even.append(l[i])
      else:
        l_odd.append(l[i])

    return l_even, l_odd
````

````python
# Tests 5 points: split_list()
def assert_eq(a, b):
    assert a == b, "{} != {}".format(a, b)

assert_eq(split_list([1, 2, 3, 4]), ([1, 3], [2, 4]))
assert_eq(split_list(['a', 'b', 'c', 'd', 'e']), (['a', 'c', 'e'], ['b', 'd']))
assert_eq(split_list([]), ([], []))


print("All tests pass!")
````

## Question 2
Write a function `merge_lists` that undoes the previous split, so that:

```
merge_lists(['a', 'c', 'e'], ['b', 'd'])
```

returns

```
['a', 'b', 'c', 'd', 'e']
```

````python
def merge_lists(l1, l2):
    """Merge the lists l1 and l2, picking alternating elements
    from the lists, starting from l1."""
    # You don't need to worry about lists of the wrong lengths.
    assert len(l2) <= len(l1) <= len(l2) + 1
    ### YOUR SOLUTION HERE
    l3 = []
    for i in range(max(len(l1), len(l2))):
      if i < len(l1):
        l3.append(l1[i])
      if i < len(l2):
        l3.append(l2[i])
    return l3

````

````python
# Tests 5 points: merge_lists()
def assert_eq(a, b):
    assert a == b, "{} != {}".format(a, b)

# Same length.
assert_eq(merge_lists(['a', 'b', 'c'], ['d', 'e', 'f']),
          ['a', 'd', 'b', 'e', 'c', 'f'])

# Longer by one.
assert_eq(merge_lists(['a', 'b', 'c'], ['d', 'e']),
          ['a', 'd', 'b', 'e', 'c'])

print("All tests pass!")
````

## Question 2: List Compression

Consider a list containing repeated elements, such as:

    ['a', 'a', 'b', 'b', 'b', 'c', 'd', 'd', 'd', 'd']

To save space, it may be convenient to represent the list as a list of pairs *(# repetitions, element),* rather than as a list of elements.\
In this representation, the list above would be:

    [(2, 'a'), (3, 'b'), (1, 'c'), (4, 'd')]

meaning:

* 2 repetitions of `'a'`
* 3 repetitions of `'b'`
* 1 repetition of `'c'`
* 4 repetitions of `'d'`

Precisely, we say that a *compressed list* is a list consisting of pairs $(n, a)$, each indicating that element $a$ is repeated $n$ **consecutive** times.  

Please write two functions, `compress()` and `decompress()`, that compress and decompress a list to/from such a representation.  Of course, you should produce efficient representations, so that:

* There are no $(0, a)$ elements in the compressed list (zero occurrences don't matter),
* If the list contains $\ldots, (n, a), (m, b), \ldots$, then $a \neq b$, or else you could replace the two elements by $(n+m, a)$.

The second point implies that you should not compress the list `['a', 'a', 'b']` into `[(1, 'a'), (1, 'a'), (1, 'b')]`; you should instead compress it into `[(2, 'a'), (1, 'b')]`.

Given an empty list, the `compress()` function should return an empty list.

````python
def compress(l):
    """Compresses a list into a list of pairs (num_repetitions, element)."""
    ### YOUR SOLUTION HERE
    if not l:
      return []

    result = []
    count = 1
    prev = l[0]

    for i in range(1, len(l)):
      if l[i] == prev:
        count += 1
      else:
        result.append((count, prev))
        prev = l[i]
        count = 1
    result.append((count, prev))

    return result
````

````python
# Tests 5 points: compress(), simple cases

def assert_eq(a, b):
    assert a == b, "{} != {}".format(a, b)

# Let us test simple cases first. These tests should pass (not print an error):
assert_eq(compress([]), [])
assert_eq(compress(['a']), [(1, 'a')])
assert_eq(compress(['a', 'a']), [(2, 'a')])


print("All tests pass!")
````

````python
# Tests 5 points: compress(), complex cases

def assert_eq(a, b):
    assert a == b, "{} != {}".format(a, b)

# Now some more complicated cases.
assert_eq(compress([1, 2, 2, 3, 3, 3, 0, 'cat']), [(1, 1), (2, 2), (3, 3), (1, 0), (1, 'cat')])

print("All tests pass!")
````

````python
def decompress(c):
    """Decompresses the compressed list c in to the list of its elements."""
    ### YOUR SOLUTION HERE
    result = []
    for count, elem in c:
      result.extend([elem] * count)
    return result
````

````python
# Tests 5 points: decompress(), simple cases

def assert_eq(a, b):
    assert a == b, "{} != {}".format(a, b)

# Here are some simple tests.
assert_eq(decompress([]), [])
assert_eq(decompress([(1, 'a')]), ['a'])
assert_eq(decompress([(2, 'a')]), ['a', 'a'])
print("All tests pass!")
````

## Question 3: Function composition via dictionaries

A dictionary implements a function, as a lookup table: for instance, if you have a function $f$ that has values

$$
f(1) = 4, \quad f(2) = 7, \quad f(5) = 7
$$

You can implement it via the dictionary:

```
f = {1: 4, 2: 7, 5: 7}
```

Given two functions `f` and `g`, represented as dictionaries, write a function called `compose(f, g)` that returns a dictionary `h`, defined by:

$$
ℎ(𝑥)=𝑔(𝑓(𝑥))
$$

That is, the dictionary `h` associates to each $𝑥$ the value  $𝑔(𝑓(𝑥))$. This value can be computed by first looking up  $𝑥$ in `f`, and then by looking up the result in `g`. If any lookup fails, the function is undefined, and there is no entry in the result dictionary.

Given the function `f` in the example above, if

```
g = {7: 8, 9: 10}
```

then `compose(f, g)` should return the dictionary:

```
{2: 8, 5: 8}
```

because

* `f[1]` is `4`, but `4` is not a key in `g`. So there is no entry for the key `1` in the result dictionary
* `f[2]` is `7`, and `g[7]` is `8`
* `f[5]` is `7`, and `g[7]` is `8`

````python
def compose(f, g):
    ### YOUR SOLUTION HERE
    h = {}
    for x in f:
      if f[x] in g:
        h[x] = g[f[x]]
    return h

````

````python
# Tests 10 points: compose()

def assert_eq(a, b):
    assert a == b, "{} != {}".format(a, b)

f = {1: 4, 2: 7, 5: 7}
g = {7: 8, 9: 10}

assert_eq(compose(f, g), {2: 8, 5: 8})
assert_eq(compose(g, f), dict())

f = {2: 3, 3: 2}
assert_eq(compose(f, f), {2: 2, 3: 3})

f = {1: 2, 3: 4, 4: 5}
g = {2: 3, 5: 1}
assert_eq(compose(f, g), {1: 3, 4: 1})


print("All tests pass!")
````

## Question 4: Representing segments

We are going to define a class to represent segments (or if you prefer, intervals in the real line). The `Segment` class has two attributes: `self.l` and `self.r`, representing the two endpoints.

```
class Segment(object):

    def __init__(self, a, b):
        self.l = min(a, b) # Left endpoint
        self.r = max(a, b) # Right endpoint
```

We can then create a segment like the following:

```
s = Segment(4, 6)
```
so that `s` is a segment from 4 to 6.

Write a new method called `intersect(self, other)` for the `Segment` class. Given a segment `s` and a segment `t`, the **intersection** `s.intersect(t)` should return
* `None` if the segments `s` and `t` do not intersect;
* a new `Segment` object representing the intersection of `s` and `t`, if they do intersect.

For example, given
```
s = Segment(4, 6)
t = Segment(5, 10)
r = Segment(6, 10)
q = Segment(7, 10)
```

then:

```
s.intersect(t) returns Segment(5, 6)
t.intersect(s) returns Segment(5, 6)

s.intersect(r) returns Segment(6, 6) # It's ok if the endpoints coincide!
r.intersect(s) returns Segment(6, 6)

s.intersect(q) returns None
q.intersect(s) returns None
```

````python
class Segment(object):

    def __init__(self, a, b):
        self.l = min(a, b) # Left endpoint
        self.r = max(a, b) # Right endpoint

    def intersect(self, other):
        ### YOUR SOLUTION HERE
        new_left = max(self.l, other.l)
        new_right = min(self.r, other.r)

        if new_left > new_right:
          return None
        return Segment(new_left, new_right)


````

````python
# Tests 10 points: tests for intersection

Segment.__repr__ = lambda self: "[{}, {}]".format(self.l, self.r)
Segment.__eq__ = lambda self, other: other is not None and self.l == other.l and self.r == other.r

def assert_eq(a, b):
    assert a == b, "{} != {}".format(a, b)

s = Segment(4, 6)
t = Segment(5, 10)
r = Segment(6, 10)
q = Segment(7, 10)

assert_eq(s.intersect(t), Segment(5, 6))
assert_eq(t.intersect(s), Segment(5, 6))

assert_eq(s.intersect(r), Segment(6, 6)) # It's ok if the endpoints coincide!
assert_eq(r.intersect(s), Segment(6, 6))

assert_eq(s.intersect(q), None)
assert_eq(q.intersect(s), None)

assert_eq(Segment(28, 60).intersect(Segment(2, 40)), Segment(28, 40))

# You should not modify the segments
assert_eq(s, Segment(4, 6))
assert_eq(t, Segment(5, 10))
assert_eq(r, Segment(6, 10))
assert_eq(q, Segment(7, 10))


print("All tests pass!")
````
